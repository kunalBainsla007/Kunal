```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
match_win = df['match_winner'].value_counts()
```

```python
sns.barplot(y=match_win.index, x=match_win.values, palette="rainbow")
plt.title("Match Winner")
plt.show
```

```python
df.head()
```

```python
sns.countplot(x=df['toss_decision'], palette="rainbow")


```

```python
count = df[df["match_winner"] == df["toss_winner"]]["match_id"].count()

percentage = (count * 100)/df.shape[0]
percentage.round(2)
```

```python
sns.countplot(x = df["won_by"])
plt.title("Won By")
plt.show()
```

```python
count = df["player_of_the_match"].value_counts().head(10)
count
```

```python
sns.barplot(y=count.index, x=count.values, palette="rainbow")
plt.title("Player Of The Match")
plt.show()
```

```python
high = df.groupby("top_scorer")["highscore"].sum().sort_values(ascending=False).head(10)
high
```

```python
high.plot(kind = "bar")
plt.title("High Score")
```

```python
df["highest_wickets"] = df["best_bowling_figure"].apply(lambda x :x.split("--")[0])
```

```python
df
```

```python
df["highest_wickets"] = df["highest_wickets"].astype(int)
```

```python
top_bowler = df.groupby("best_bowling")["highest_wickets"].sum().sort_values(ascending=False)
top_bowler
```

```python
venue_count = df["venue"].value_counts()
venue_count
```

```python
sns.barplot(x = venue_count.values, y = venue_count.index, palette="rainbow")
plt.title("Venue Count")
plt.show()
```
